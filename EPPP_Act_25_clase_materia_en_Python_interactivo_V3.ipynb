#### Clase Materia con tres atributos (nombre, clave, créditos)
#### funciones leerdatos(), mostrardatos()
#### que sea interactivo --> los input
#### Crear dos objetos (materia1, materia2).
#### no usar while true, if, for, lista y diccionario
![link text](https://raw.githubusercontent.com/SantiagoMedrano128/Clase-Materia-act-25-Santiago-0103/refs/heads/main/infoclasemateria.png)




```python
class Materia:
    def __init__(self, nombre, clave, creditos):
      self.nombre = nombre
      self.clave = clave
      self.creditos = creditos

    def leerdatos(self):
      self.nombre = input("Ingrese el nombre de la materia: ")
      self.clave = input("Ingrese la clave de la materia: ")
      self.creditos = input("Ingresa la cantidad de creditos: ")

    def mostrardatos(self):
      print("Nombre: ", self.nombre)
      print("Clave: ", self.clave)
      print("Creditos: ", self.creditos)

# Crear 2 objetos materia1, materia2 después de la definición de la clase
print("Materia 1")
materia1 = Materia("","", "")
materia1.leerdatos()
materia1.mostrardatos()
print("-*-*-*-*-*-*-*-*-*-*-*-*")
print("Materia 2")
materia2 = Materia("","", "")
materia2.leerdatos()
materia2.mostrardatos()
print("-*-*-*-*-*-*-*-*-*-*-*-*")
print("Programa realizado por = Santiago Medrano NL = 37")
```
